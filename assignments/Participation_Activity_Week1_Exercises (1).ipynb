## Exercise 7: Scaling data
### min-max scaling

```python
import numpy as np
min1 = np.min(salaries)
max1 = np.max(salaries)
range1 = max1 - min1
scaled = [(value - min1) / range1 for value in salaries]
print(scaled)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
standardized = [(x - mean) / stdev(salaries) for x in salaries]
standardized[:100]
cov=np.cov(scaled, standardized)
print(cov)
```

### Pearson correlation coefficient ($\rho$)

```python
from statistics import stdev
scaled1=stdev(scaled)
standardized1=stdev(standardized)
print(scaled1)
print(standardized1)
pearson=cov[0,1]/(scaled1*standardized1)
print(pearson)
```

Because the Pearson correlation coefficient can't be a matrix, also "The variance of scaled" and "The variance of standardized" is meanlingless. We need to pull the cov[0,1]out to calculate the right one.

$$\rho_{X, Y} = \frac{cov(X, Y)}{s_X s_Y}$$
