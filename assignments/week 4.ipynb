```python
import pandas as pd
data = pd.read_csv('/content/drive/MyDrive/MSSP607/Codesamples/parsed.csv')
```

```python
# a. Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
japan=data[data["place"].str.contains("Japan", na=False)]
japan_mb=japan[japan['magType']=='mb']
japan_mb["mag"].quantile(0.95)
```

```python
#b. Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.
Indonesia=data[data['place'].str.contains('Indonesia',na=False)]
Indonesia_tsunamis=Indonesia[Indonesia['tsunami']==1]
Percentage=len(Indonesia_tsunamis)/len(Indonesia)*100
print(Percentage)
```

```python
#c. Get summary statistics for earthquakes in Nevada
Nevada=data[data['place'].str.contains('Nevada',na=False)]
Nevada.describe()
```

```python
#d. Add a column to the dataframe indicating whether or not the earthquake happened
#in a country or US state that is on the Ring of Fire.
#Use Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico (be sure not to select New Mexico),
#Japan, Philippines, Indonesia, New Zealand, Antarctica (look for the Antarctic), Canada, Fiji, Alaska, Washington,
# California, Russia, Taiwan, Tonga, and the Kermadec Islands).
data['ring_of_fire'] = data.parsed_place.str.contains(r'|'.join([
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
    'Guatemala', '^Mexico', 'Japan', 'Philippines', 'Indonesia',
    'New Zealand', 'Antarctic', 'Canada', 'Fiji',
    'Alaska', 'Washington', 'California', 'Russia',
    'Taiwan', 'Tonga', 'Kermadec Islands'
]))

data[['ring_of_fire', 'parsed_place']]
```

```python
#e. Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them,
filtered_data = data[(data['ring_of_fire'] == True) & (data['type'] == 'earthquake')]
total_inside = data[(data['ring_of_fire'] == True) & (data['type'] == 'earthquake')].type.count()
total_outside = data[(data['ring_of_fire'] == False) & (data['type'] == 'earthquake')].type.count()
print('The total number of earthquakes in the Ring of Fire is: ', total_inside)
print('The total number of earthquakes outside the Ring of Fire is: ', total_outside)
```
