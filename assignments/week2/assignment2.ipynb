```python
food = ['apples', 'bananas', 'tofu', 'cats']
def spam1(food):
    first_part = ', '.join(food[:-1])
    last_part = food[-1]
    result = first_part + ', and ' + last_part
    return result
print(spam1(food))
```

.join() → join / connect

First, food[:-1] selects all items except the last one, while food[-1] selects the last item. join() combines the first items with commas. Then, the code adds ", and " before the last item. Finally, return returns the complete string, and print() displays it.
