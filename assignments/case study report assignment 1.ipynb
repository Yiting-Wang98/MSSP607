**step1:Parental Education Data Cleaning and Coding**


```python
saber11_clean =saber11.copy()
saber11_clean["EDU_FATHER"] = saber11_clean["EDU_FATHER"].str.strip()
saber11_clean["EDU_MOTHER"] = saber11_clean["EDU_MOTHER"].str.strip()
edu_score = {"Ninguno": 1,
    "Incomplete primary": 2,
    "Complete primary": 3,
    "Incomplete Secundary": 4,
    "Complete Secundary": 5,
    "Incomplete technical or technological": 6,
    "Complete technique or technology": 7,
    "Incomplete Professional Education": 8,
    "Complete professional education": 9,
    "Postgraduate education": 10,}
saber11_clean["EDU_FATHER_NUM"] = saber11_clean["EDU_FATHER"].map(edu_score)
saber11_clean["EDU_MOTHER_NUM"] = saber11_clean["EDU_MOTHER"].map(edu_score)
```

```python
import statsmodels.api as sm
# Keep only variables needed for the regression
reg_data = saber11_clean[
    ["G_SC", "EDU_MOTHER_NUM", "EDU_FATHER_NUM"]].dropna()

# Independent variables
X = reg_data[["EDU_MOTHER_NUM", "EDU_FATHER_NUM"]]

# Add intercept
X = sm.add_constant(X)

# Dependent variable
y = reg_data["G_SC"]

# Run multiple OLS regression
model = sm.OLS(y, X).fit()

# Display results
print(model.summary())
```

Line 1: dropna temporarily filters out the rows where the education score is empty (these are the 0 and Not sure entries), because pearsonr raises an error when it encounters missing values. This is only used for this step; saber11_clean itself is unchanged.
Line 2: The output contains statistic (the correlation coefficient) and pvalue (the p-value).

```python
valid = saber11_clean.dropna(subset=["STRATUM", "G_SC"])
groups = [
    group["G_SC"].values
    for _, group in valid.groupby("STRATUM")]

stats.f_oneway(*groups)
```

1.Group the data: groupby("STRATUM") separates students into different socioeconomic strata.
2.Select each group: for _, group goes through each stratum one at a time.
3.Select G_SC: group["G_SC"] extracts the Global Score for each stratum.
4.Convert to arrays: .values converts the scores into arrays for the ANOVA test.
5.Store the groups: groups = [...] collects the G_SC scores from all strata into one list.

**4.Is the relationship between parental education and academic performance weaker at the university stage than at the secondary-school stage?**

```python
s11_subjects = ["MAT_S11", "CR_S11", "CC_S11", "BIO_S11", "ENG_S11"]
saber11_clean["S11_AVG"] = saber11_clean[s11_subjects].mean(axis=1)
```

```python
print(saber11_clean[["MAT_S11", "CR_S11", "CC_S11", "BIO_S11", "ENG_S11", "S11_AVG"]])
```

```python
saber11_clean.groupby("EDU_FATHER_NUM")["S11_AVG"].agg(["count", "mean", "median"]).round(1)
```

```python
saber11_clean.groupby("EDU_MOTHER_NUM")["S11_AVG"].agg(["count", "mean", "median"]).round(1)
```

```python
valid = saber11_clean.dropna(subset=["EDU_FATHER_NUM"])
stats.pearsonr(valid["EDU_FATHER_NUM"], valid["S11_AVG"])
```

```python
valid = saber11_clean.dropna(subset=["EDU_MOTHER_NUM"])
stats.pearsonr(valid["EDU_MOTHER_NUM"], valid["S11_AVG"])
```

The results show a positive relationship between parental education and academic performance at both the secondary-school and university stages. At the secondary-school stage, father's education was positively correlated with students' average Saber 11 score (r = .315, p < .001), while mother's education was also positively correlated with performance (r = .312, p < .001). At the university stage, the correlations were slightly weaker, with father's education correlated with the Global Score (r = .261, p < .001) and mother's education correlated with the Global Score (r = .266, p < .001). Overall, the correlations suggest that the relationship between parental education and academic performance was somewhat stronger at the secondary-school stage than at the university stage.
