
In the context of downloading data from Yahoo Finance using the yfinance library, the period parameter defines the range of data to download and can have the following values:

*    1d, 5d: data for the past 1 or 5 days

*   1mo, 3mo, 6mo: data for the past 1, 3, or 6 months
*   1y, 2y, 5y, 10y, ytd: data for the past 1, 2, 5, or 10 years, or year to date.


*   max: data for the maximum available period.


These values allow you to specify the time frame for which you want to retrieve historical data for a given stock.

#Task 1: Read into seperate data frame the following tickers: Apple, Facebook, Google, and Netflix; use Yahoo finance.

```python
tickers = {
    'Apple': 'AAPL',
    'Facebook': 'META',
    'Google': 'GOOGL',
    'Netflix': 'NFLX'}

Apple = yf.Ticker("AAPL").history(period="max")
Facebook = yf.Ticker("META").history(period="max")
Google = yf.Ticker("GOOGL").history(period="max")
Netflix = yf.Ticker("NFLX").history(period="max")

print("Apple Data")
display(Apple.head())

print("Facebook Data")
display(Facebook.head())

print("Google Data")
display(Google.head())

print("Netflix Data")
display(Netflix.head())
```

#Task 2:  Join the individual stock data frames into a single data frame.

```python
# Add ticker s
Apple['Ticker'] = 'AAPL'
Facebook['Ticker'] = 'META'
Google['Ticker'] = 'GOOGL'
Netflix['Ticker'] = 'NFLX'

# Combine
combined_df = pd.concat([
    Apple,
    Facebook,
    Google,
    Netflix])

# Reset index
combined_df = combined_df.reset_index()

# Display combined data frame
display(combined_df.head())

# Check the data
print(combined_df.shape)
print(combined_df['Ticker'].value_counts())
```

# Task 3: Create a visualization of the following financial metrics for each stock:  quickRatio, currentRatio, totalRevenue, debtToEquity, revenuePerShare, returnOnAssets, returnOnEquity, grossProfits


```python
# Define metrics
financial_metrics = [
    'quickRatio',
    'currentRatio',
    'totalRevenue',
    'debtToEquity',
    'revenuePerShare',
    'returnOnAssets',
    'returnOnEquity',
    'grossProfits']
financial_data = []

for company, ticker in tickers.items():

    stock = yf.Ticker(ticker)

    info = stock.info

    row = {'Company': company, 'Ticker': ticker}

    for metric in financial_metrics:
        row[metric] = info.get(metric, np.nan)

    financial_data.append(row)

# Create a financial metrics data frame
financial_df = pd.DataFrame(financial_data)

display(financial_df)
```

```python
# Visualize each  metric
for metric in financial_metrics:

    plt.figure(figsize=(8, 5))

    sns.barplot(
        data=financial_df,
        x='Company',
        y=metric
    )

    plt.title(f'{metric} Comparison Across Companies')
    plt.xlabel('Company')
    plt.ylabel(metric)
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```

#Task 4: Create a single boxplot for the Closing price for each stock.


```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=combined_df,
    x='Ticker',
    y='Close')

plt.title('Closing Price Distribution Across Four Companies')
plt.xlabel('Company')
plt.ylabel('Closing Price')
plt.tight_layout()
plt.show()
```

# Task 5: Create a daily trading volume visualization for the year 2023 for each stock.


```python

volume_data = {}

for company, ticker in tickers.items():

    stock = yf.Ticker(ticker)

    df = stock.history(
        start='2023-01-01',
        end='2024-01-01')

    volume_data[company] = df

# Create daily trading volume visualization
plt.figure(figsize=(14, 7))

for company, df in volume_data.items():

    plt.plot(
        df.index,
        df['Volume'],
        label=company)

plt.title('Daily Trading Volume of Four Companies in 2023')
plt.xlabel('Date')
plt.ylabel('Trading Volume')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
