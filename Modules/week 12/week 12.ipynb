```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('/content/MSSP607/Modules/week 12/Student Performance Data (2).csv')
df.columns = (df.columns.str.strip().str.lower()
              .str.replace(' ', '_').str.replace('/', '_'))
print(df.columns.tolist())   # check the column names here
df.head()
```

# H0: the mean scores of males with and without the preparation course are equal
# H1: the mean scores are different (two-tailed), alpha = 0.05

```python
males = df[df['gender'] == 'male']
prep   = males[males['test_preparation_course'] == 'completed']
noprep = males[males['test_preparation_course'] == 'none']

print('n (prep):', len(prep), ' n (no prep):', len(noprep))
print('mean total score (prep):   ', round(prep['total_score'].mean(), 2))
print('mean total score (no prep):', round(noprep['total_score'].mean(), 2))
```

```python

t, p = stats.ttest_ind(prep['total_score'], noprep['total_score'], equal_var=False)
print('t =', round(t, 2), ' p =', p)

if p < 0.05:
    print('Reject H0: there is a significant difference.')
else:
    print('Fail to reject H0: no significant difference.')
```

```python
for s in ['math_score', 'reading_score', 'writing_score']:
    t, p = stats.ttest_ind(prep[s], noprep[s], equal_var=False)
    print(s, '| prep:', round(prep[s].mean(), 1), '| no prep:', round(noprep[s].mean(), 1),
          '| t =', round(t, 2), '| p =', p)
```

# H0: males and females have the same mean total score

```python

male = df[df['gender'] == 'male']['total_score']
female = df[df['gender'] == 'female']['total_score']

t, p = stats.ttest_ind(male, female, equal_var=False)
print('male mean:', round(male.mean(), 2), ' female mean:', round(female.mean(), 2))
print('t =', round(t, 2), ' p =', p)

if p < 0.05:
    print('Reject H0: there is a significant difference.')
else:
    print('Fail to reject H0: no significant difference.')
```

**2. Is there a statistical difference between males who took the preparation course and others?**

Yes. Using Welch's t-test, males who completed the preparation course (n = 174) scored significantly higher than males who did not (n = 308) on every measure:

- Total score: 212.34 vs. 189.13 (t = 6.18, p ≈ 1.7×10⁻⁹)
- Math: 72.3 vs. 66.7 (t = 4.21, p ≈ 3.3×10⁻⁵)
- Reading: 70.2 vs. 62.8 (t = 5.77, p ≈ 1.8×10⁻⁸)
- Writing: 69.8 vs. 59.6 (t = 8.14, p ≈ 6.2×10⁻¹⁵)

All p-values are below 0.05, so we reject H0.

**3. Are there any other differences that are of interest?**

Yes. Females scored significantly higher than males on total score (208.71 vs. 197.51, t = -4.18, p ≈ 3.2×10⁻⁵), so we reject H0 that the two means are equal.

Note: the data are observational, so these results show an association, not that the preparation course causes higher scores.
