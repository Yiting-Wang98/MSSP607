```python
import requests, bs4, re

search_word = 'kitten'
max_images = 10
save_folder = 'Modules/week 9/week 9 photo'

res = requests.get('https://www.flickr.com/search/?text=' + search_word)
res.raise_for_status()
print(res.url)

soup = bs4.BeautifulSoup(res.text, 'html.parser')
img_tags = soup.find_all('img', src=re.compile('staticflickr'))
print('Number of image tags found:', len(img_tags))

count = 0
for tag in img_tags[:max_images]:
    link = tag['src']
    if link.startswith('//'):
        link = 'https:' + link
    print(link)
    img = requests.get(link)
    try:
        img.raise_for_status()
    except Exception as exc:
        print('There was a problem: %s' % (exc))
        continue
    count = count + 1
    photofile = open(save_folder + search_word + str(count) + '.jpg', 'wb')
    for chunk in img.iter_content(100000):
        photofile.write(chunk)
    photofile.close()

print('Done. Images saved:', count)
```
